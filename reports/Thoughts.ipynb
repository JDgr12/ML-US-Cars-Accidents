### Caracteristicas:

- **Problema de Clasificación:** **El objetivo es predecir la severidad del accidente**, que se encuentra en una escala del 1 al 4

Resolverlo como un problema binario suele ser mas directo y util para rapida accion (identificar el peor de los casos)
- **El volumne de los datos:** El dataset completo cuenta con aproximadamente con 7.7 millones de registros. Esto es bastante pesado para procesar. Es permitido usar la versionreducida de 1 millon de filas
- **Las variables:** Tienen 47 columnas en contexto (clima, ubicacion, geografia, puntos de interes, tiempo) sin necesidad de saber sobre los vehiculos. **Trabajo lidiando con los nulos, las variables de fecha/hora, y variables booleanas de infrastructura**
